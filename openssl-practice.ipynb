## 1. Install OpenSSL Software (MacOS)

OpenSSL is a robust, full-featured open-source toolkit for the Transport Layer Security (TLS) and Secure Sockets Layer (SSL) protocols. It's used for generating cryptographic keys, certificates, and performing various cryptographic operations.

On macOS, you can install OpenSSL using Homebrew, a popular package manager for macOS.

```python
!brew install openssl
```

## 2. Generate Private RSA Key

A private RSA key is the foundation of asymmetric cryptography. This key must be kept secret and is used to:
- Decrypt messages encrypted with the corresponding public key
- Sign data to prove authenticity and non-repudiation

The command generates a 2048-bit RSA private key, which is a standard secure key size. The key is saved in PEM (Privacy Enhanced Mail) format, a widely supported text-based format.

```python
!openssl genrsa -out workspace/private_key.pem 2048
```

## 3. Check Generated Key

This command displays the detailed contents of the private key including modulus, exponents, and prime factors to verify it was created correctly.

```python
!openssl rsa -in workspace/private_key.pem -text -noout
```

## 4. Generate Public Key

The public key is derived from the private key and can be shared with others to encrypt messages or verify signatures.

```python
!openssl rsa -in workspace/private_key.pem -outform PEM -pubout -out workspace/public_key.pem
```

## 5. Encrypt Message with Public Key

This encrypts a message using the public key, ensuring only the holder of the corresponding private key can decrypt it.

```python
!openssl pkeyutl -encrypt -inkey workspace/public_key.pem -pubin -in input/secret-message.txt -out workspace/encrypted.txt
```

## 6. Decrypt Message with Private Key

This decrypts the encrypted message using the private key, allowing only the legitimate owner to read the original content.

```python
!openssl pkeyutl -decrypt -inkey workspace/private_key.pem -in workspace/encrypted.txt -out workspace/decrypted.txt
```

## 7. Compare Decrypted and Original Messages

This compares the decrypted message with the original secret message to verify that the encryption/decryption process worked correctly and the messages match perfectly.

```python
!echo "=== Decrypted Message ===" && cat workspace/decrypted.txt && echo -e "\n=== Original Message ===" && cat input/secret-message.txt && echo -e "\n=== Comparison ===" && diff workspace/decrypted.txt input/secret-message.txt && echo "Messages match!" || echo "Messages do not match!"
```
